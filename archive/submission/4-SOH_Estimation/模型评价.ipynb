```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
plt.style.use('default')
plt.rcParams['font.sans-serif']=['SimHei']
plt.rcParams['axes.unicode_minus']=False 
import pickle
from sklearn.preprocessing import StandardScaler
from itertools import combinations_with_replacement
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
cm = 1/2.54
# 字号
fontsize = 8
seed = 2025
```

```python
def model_eval(y_true, y_pred):
    mse = mean_squared_error(y_true, y_pred)*100
    mae = mean_absolute_error(y_true, y_pred)*100
    r2 = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))*100
    return rmse, mse, mae, r2 
```

```python
# 遍历结果文件夹下的CSV文件，计算各个模型的评价指标
def get_metrics(result_folder):
    files = os.listdir(result_folder)
    metrics = pd.DataFrame()
    for file in files:
        if file.endswith('.csv'):
            df = pd.read_csv(os.path.join(result_folder, file))
            y_true = df['real']
            y_pred = df['pred']
            rmse, mse, mae, r2 = model_eval(y_true, y_pred)
            df = pd.DataFrame({'model': [file.split('_')[1].split('.')[0]], 
                                    'rmse': [rmse], 'mse': [mse], 
                                    'mae': [mae], 'r2': [r2]})
            metrics = pd.concat([metrics, df])
    return metrics
```

```python
folder_CC = r'D:\My_program\long\code\4-SOH_Estimation\models\CC'
folder_CV = r'D:\My_program\long\code\4-SOH_Estimation\models\CV'
folder_RE = r'D:\My_program\long\code\4-SOH_Estimation\models\RE'
metrics_CC = get_metrics(folder_CC)
metrics_CV = get_metrics(folder_CV)
metrics_RE = get_metrics(folder_RE)
```

```python
folder = r'D:\My_program\long\code\4-SOH_Estimation\人工特征模型结果'
metrics_CC.to_csv(os.path.join(folder, '人工特征_CC.csv'), index=False)
metrics_CV.to_csv(os.path.join(folder, '人工特征_CV.csv'), index=False)
metrics_RE.to_csv(os.path.join(folder, '人工特征_RE.csv'), index=False)
```
