```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.layers import Input, Dense, Concatenate
from tensorflow.keras.models import Model
import shap

# 生成合成数据
np.random.seed(42)
n_samples = 1000

# 输入1：数值特征（2个特征）
x1 = np.random.randn(n_samples, 2)
# 输入2：数值特征（3个特征）
x2 = np.random.randn(n_samples, 3)
# 目标值（二分类）
y = np.where((x1[:, 0] > 0.5) & (x2[:, 1] < -0.2), 1, 0)

# 构建双输入神经网络
def build_two_input_model():
    # 输入层
    input1 = Input(shape=(2,), name='input1')
    input2 = Input(shape=(3,), name='input2')
    
    # 分支处理
    dense1 = Dense(8, activation='relu')(input1)
    dense2 = Dense(8, activation='relu')(input2)
    
    # 合并
    merged = Concatenate()([dense1, dense2])
    
    # 全连接层
    output = Dense(1, activation='sigmoid')(Dense(4, activation='relu')(merged))
    
    model = Model(inputs=[input1, input2], outputs=output)
    model.compile(optimizer='adam', loss='binary_crossentropy')
    return model

# 训练模型
model = build_two_input_model()
model.fit([x1, x2], y, epochs=50, batch_size=32, verbose=0)

# SHAP解释器设置
background = [x1[:100], x2[:100]]  # 背景数据集
explainer = shap.DeepExplainer(model, background)

# 计算测试样本的SHAP值
test_samples = 10
x1_test = x1[:test_samples]
x2_test = x2[:test_samples]
shap_values = explainer.shap_values([x1_test, x2_test])

# 可视化解释结果
print("SHAP values for Input 1 features:")
shap.summary_plot(shap_values[0], x1_test, feature_names=['Feature1', 'Feature2'])
print("\nSHAP values for Input 2 features:")
shap.summary_plot(shap_values[1], x2_test, feature_names=['Feature3', 'Feature4', 'Feature5'])
```
