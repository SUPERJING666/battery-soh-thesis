#### CV数据

```python
with open(r'D:\My_program\long\code\5-fusion\data_CV.pkl', 'rb') as file:
    Ctrain_CV, Cval_CV, Ctest_CV, Htrain_CV, Hval_CV, Htest_CV = pickle.load(file)
    file.close()
```

```python
def get_inut(Curve, HI):
    x = pd.concat([Curve.iloc[:, :-3], HI.iloc[:, :-3]], axis=1)
    y = HI.iloc[:, -3:]
    return x, y
```

```python
train_x, train_y = get_inut(Ctrain_CV, Htrain_CV)
val_x, val_y = get_inut(Cval_CV, Hval_CV)
test_x, test_y = get_inut(Ctest_CV, Htest_CV)
print(train_x.columns)
```

```python
# 标准化
scaler = StandardScaler()
train_x = scaler.fit_transform(train_x)
val_x = scaler.transform(val_x)
test_x = scaler.transform(test_x)
with open (r'D:\My_program\long\code\8-explain\data\CV_early.pkl','wb') as file:
    pickle.dump([train_x, train_y, val_x, val_y, test_x, test_y], file)
    file.close()
```

#### CC和RE数据

```python
def get_xy(Cdata, Hdata):
    input_1 = Cdata.iloc[:, :-3]
    input_2 = Hdata.iloc[:, :-3]
    output = Cdata.iloc[:, -3:]
    return input_1, input_2, output
# 归一化函数
def data_normlization(train_data, val_data, test_data):
    scaler = StandardScaler()
    train_data = scaler.fit_transform(train_data)
    val_data = scaler.transform(val_data)
    test_data = scaler.transform(test_data)
    return train_data, val_data, test_data
```

```python
# 加载数据集，Ctrain_CC表示曲线训练集的CC数据，Htrain_CC表示曲线训练集的HI数据
with open(r'D:\My_program\long\code\5-fusion\data_CC.pkl', 'rb') as file:
    Ctrain_CC, Cval_CC, Ctest_CC, Htrain_CC, Hval_CC, Htest_CC = pickle.load(file)
    file.close()
# 划分输入输出
train_x1, train_x2, train_y = get_xy(Ctrain_CC, Htrain_CC)
val_x1, val_x2, val_y = get_xy(Cval_CC, Hval_CC)
test_x1, test_x2, test_y = get_xy(Ctest_CC, Htest_CC)
# 归一化
train_x1, val_x1, test_x1 = data_normlization(train_x1, val_x1, test_x1)
train_x2, val_x2, test_x2 = data_normlization(train_x2, val_x2, test_x2)
with open (r'D:\My_program\long\code\8-explain\data\CC_late.pkl','wb') as file:
    pickle.dump([train_x1, train_x2, train_y, val_x1, val_x2, val_y, test_x1, test_x2, test_y], file)
    file.close()
```

```python
# 加载数据集
with open(r'D:\My_program\long\code\5-fusion\data_RE.pkl', 'rb') as file:
    Ctrain_RE, Cval_RE, Ctest_RE, Htrain_RE, Hval_RE, Htest_RE = pickle.load(file)
    file.close()
# 划分输入输出
train_x1, train_x2, train_y = get_xy(Ctrain_RE, Htrain_RE)
val_x1, val_x2, val_y = get_xy(Cval_RE, Hval_RE)
test_x1, test_x2, test_y = get_xy(Ctest_RE, Htest_RE)
print(train_x2.columns)
# 归一化
train_x1, val_x1, test_x1 = data_normlization(train_x1, val_x1, test_x1)
train_x2, val_x2, test_x2 = data_normlization(train_x2, val_x2, test_x2)
with open (r'D:\My_program\long\code\8-explain\data\RE_late.pkl','wb') as file:
    pickle.dump([train_x1, train_x2, train_y, val_x1, val_x2, val_y, test_x1, test_x2, test_y], file)
    file.close()
```
