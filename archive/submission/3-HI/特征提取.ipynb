```python
with open(r'D:\My_program\long\code\data\data.pkl', 'rb') as file:
    df = pickle.load(file)
# 删除含有NAN的行
df = df.dropna()
```

```python
# 特征提取函数
def HI_extract(data):
    # 统计特征，Var, Ske, Max, Min, Mean, Kur
    features = pd.DataFrame(columns=['Var', 'Ske', 'Max', 'Min', 'Mean', 'Kur'])
    for i in range(0, len(data)):
        features.loc[i, 'Var'] = np.var(data.iloc[i])
        features.loc[i, 'Ske'] = data.iloc[i].skew()
        features.loc[i, 'Max'] = np.max(data.iloc[i])
        features.loc[i, 'Min'] = np.min(data.iloc[i])
        features.loc[i, 'Mean'] = np.mean(data.iloc[i])
        features.loc[i, 'Kur'] = data.iloc[i].kurt()
    return features
```

```python
df_cc = df[[f'CC_{i}' for i in range(1, 150)]]
df_cv = df[[f'CV_{i}' for i in range(1, 150)]]
df_rest = df[[f'RE_{i}' for i in range(1, 150)]]
df_info = df[['cycle', 'capacity', 'id']]
```

```python
HI_CC = HI_extract(df_cc)
HI_CV = HI_extract(df_cv)
HI_RE = HI_extract(df_rest)
```

```python
# 合并到一个DataFrame
HI = pd.concat([HI_CC, HI_CV, HI_RE, df_info], axis=1)
# 列名
HI.columns = ['CC_Var', 'CC_Ske', 'CC_Max', 'CC_Min', 'CC_Mean', 'CC_Kur',
                'CV_Var', 'CV_Ske', 'CV_Max', 'CV_Min', 'CV_Mean', 'CV_Kur',
                'RE_Var', 'RE_Ske', 'RE_Max', 'RE_Min', 'RE_Mean', 'RE_Kur',
                'cycle', 'capacity', 'id']
# 保存数据
with open(r'D:\My_program\long\code\data\HI.pkl', 'wb') as file:
    pickle.dump(HI, file)
```
