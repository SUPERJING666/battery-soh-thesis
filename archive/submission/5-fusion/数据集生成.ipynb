```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
plt.style.use('default')
plt.rcParams['font.sans-serif']=['SimHei']
plt.rcParams['axes.unicode_minus']=False 
import pickle
import tensorflow as tf
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Model
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.layers import Conv1D,Conv2D,MaxPooling1D, MaxPooling2D, Flatten, Dense, Input, Dropout, LSTM, TimeDistributed, Reshape, BatchNormalization, Activation, concatenate
cm = 1/2.54
# 字号
fontsize = 8
seed = 2025
```

```python
# 曲线特征
with open (r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\data\data.pkl','rb') as file:
    data = pickle.load(file)
    file.close()
data = data.dropna()
# 人工特征
with open (r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\HI_CC.pkl','rb') as file:
    HI_CC = pickle.load(file)
    file.close()
with open (r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\HI_CV.pkl','rb') as file:
    HI_CV = pickle.load(file)
    file.close()
with open (r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\HI_RE.pkl','rb') as file:
    HI_RE = pickle.load(file)
    file.close()
```

```python
# 按id分为训练、验证和测试集，比例为6:2:2
id = data['id'].unique()
np.random.seed(seed)
np.random.shuffle(id)
train_id = id[:int(len(id)*0.6)]
val_id = id[int(len(id)*0.6):int(len(id)*0.8)]
test_id = id[int(len(id)*0.8):]
```

```python
# 根据id和列名获取曲线数据集
def get_curve_dataset(data, id, stage='CC'):
    dataset = data[data['id'].isin(id)]
    # 列名包括stage的列
    columns = [i for i in data.columns if stage in i]
    # columns添加'cycle', 'capacity', 'id'列
    columns = columns + ['cycle', 'capacity', 'id']
    dataset = dataset[columns]
    return dataset
Ctrain_CC = get_curve_dataset(data, train_id, 'CC')
Cval_CC = get_curve_dataset(data, val_id, 'CC')
Ctest_CC = get_curve_dataset(data, test_id, 'CC')
Ctrain_CV = get_curve_dataset(data, train_id, 'CV')
Cval_CV = get_curve_dataset(data, val_id, 'CV')
Ctest_CV = get_curve_dataset(data, test_id, 'CV')
Ctrain_RE = get_curve_dataset(data, train_id, 'RE')
Cval_RE = get_curve_dataset(data, val_id, 'RE')
Ctest_RE = get_curve_dataset(data, test_id, 'RE')
# 根据id获取人工特征数据集
def get_HI_dataset(HI, id):
    dataset = HI[HI['id'].isin(id)]
    return dataset
Htrain_CC = get_HI_dataset(HI_CC, train_id)
Hval_CC = get_HI_dataset(HI_CC, val_id)
Htest_CC = get_HI_dataset(HI_CC, test_id)
Htrain_CV = get_HI_dataset(HI_CV, train_id)
Hval_CV = get_HI_dataset(HI_CV, val_id)
Htest_CV = get_HI_dataset(HI_CV, test_id)
Htrain_RE = get_HI_dataset(HI_RE, train_id)
Hval_RE = get_HI_dataset(HI_RE, val_id)
Htest_RE = get_HI_dataset(HI_RE, test_id)
# 分别保存CC、CV、RE的训练、验证和测试集
with open(r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\data_CC.pkl','wb') as file:
    pickle.dump([Ctrain_CC, Cval_CC, Ctest_CC, Htrain_CC, Hval_CC, Htest_CC], file)
    file.close()
with open(r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\data_CV.pkl','wb') as file:
    pickle.dump([Ctrain_CV, Cval_CV, Ctest_CV, Htrain_CV, Hval_CV, Htest_CV], file)
    file.close()
with open(r'G:\boshi\1_boyi\3_ziji\My_program\long\3\code\5-fusion\data_RE.pkl','wb') as file:
    pickle.dump([Ctrain_RE, Cval_RE, Ctest_RE, Htrain_RE, Hval_RE, Htest_RE], file)
    file.close()
```
