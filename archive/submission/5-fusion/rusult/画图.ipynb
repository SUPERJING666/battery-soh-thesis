```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
plt.style.use('default')
plt.rcParams['font.sans-serif']=['SimHei']
plt.rcParams['axes.unicode_minus']=False 
import pickle
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import StandardScaler
cm = 1/2.54
# 字号
fontsize = 8
seed = 2025
```

```python
from matplotlib.ticker import MaxNLocator
def get_fig(data_folder, fig_folder):
    if not os.path.exists(fig_folder):
        os.makedirs(fig_folder)
        
    for file in os.listdir(data_folder):
        if file.endswith('.csv'):
            data = pd.read_csv(os.path.join(data_folder, file))
            plt.figure(figsize=(8*cm, 6*cm))
            plt.scatter(data['SOH'], data['pred'], s=1, c='#E64B35')
            plt.plot([data['SOH'].min(), data['SOH'].max()], [data['SOH'].min(), data['SOH'].max()], 'b--', linewidth=0.5)
            plt.xlabel('真实值', fontsize=fontsize)
            plt.ylabel('预测值', fontsize=fontsize)
            
            # 【修改点】：设置 nbins=4 (4个区间即代表 5个刻度)，这样既能固定数量，又能保证数字规整好看
            plt.gca().xaxis.set_major_locator(MaxNLocator(nbins=4))
            plt.gca().yaxis.set_major_locator(MaxNLocator(nbins=4))
            
            plt.xticks(fontsize=fontsize)
            plt.yticks(fontsize=fontsize)
            
            # 保存并关闭画布
            save_name = file.replace('.csv', '.png').split('_')[-1]
            plt.savefig(os.path.join(fig_folder, save_name), bbox_inches='tight', dpi=600)
            plt.close()
```

```python
folder_1 = r'5-fusion\rusult\early'
folder_2 = r'5-fusion\rusult\late'
folder_3 = r'5-fusion\rusult\middle'
fig_folder_1 = r'5-fusion\rusult\fig\early'
fig_folder_2 = r'5-fusion\rusult\fig\late'
fig_folder_3 = r'5-fusion\rusult\fig\middle'
get_fig(folder_1, fig_folder_1)
get_fig(folder_2, fig_folder_2)
get_fig(folder_3, fig_folder_3)
```
